```python
# import libraries relevant for this data
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# number of patient records
n = 1300

# Nominal data - generating patient_ids 
Patient_id = [f'PT{str(i).zfill(2)}' for i in range(101, 1401)]

# nominal data - generating post code for each patient
postcode_area = ['GB','HA','IT','LM','EN','SW','SE','MT','WC']
postcode_data = [f'{np.random.choice(postcode_area)}{str(i).zfill(2)}' for i in np.random.randint(1, 24, n)]

# import faker to generate names of patient and configured to the UK style names
from faker import Faker
fake = Faker("en_GB")

Patient_names = [fake.name() for i in range(n)]


# ratio data: Age of patient
age_data = np.random.randint(0, 110, n)

# Nominal data: Gender of patient
gender_category = ['Male','Female','Non-binary','Prefer not to say']
gender_data = np.random.choice(gender_category, n)

# Interval data: Date of Birth for patient in the format of year-month-day
birth_year = np.random.randint(1916, 2026, n)
birth_month = np.random.randint(1, 13, n)
birth_day = np.random.randint(1, 31, n)
birth_date = [f'{birth_year[i]}-{str(birth_month[i]).zfill(2)}-'
                     f'{str(birth_day[i]).zfill(2)}' for i in range(n)]

# create the DataFrame
Patient_data = pd.DataFrame({
    'Patient_ID': Patient_id,
    'Patient_name': Patient_names,
    'Postcode': postcode_data,
    'Age': age_data,
    'Gender':gender_data,
    'Date_of_Birth':birth_date
})

print(Patient_data.head(10))
```

```python
# Nominal data: generate department names
dept_names = ['Orthopaedics','Paediatrics','Cardiology','Neurology',
              'Oncology','Gyneacology', 'General medicine','Surgery']

# Nominal data: generate unique department ids for each department
department_id = [f'DEPT{str(i).zfill(2)}' for i in range(1, len(dept_names) + 1)]

# create DataFrame
department_data = pd.DataFrame({
    'Dept_ID':department_id,
    'Dept_name':dept_names})


print(department_data.head(10))
                   




```

```python

# number of doctors
no_of_drs = 30

# nominal data: generate unique doctor ids
doc_id = [f'DR{str(i).zfill(2)}' for i in range(1, 31)]

# nominal data: use faker to generate a name for each doctor
doctors_names = [f'Dr {fake.name()}' for i in range(no_of_drs)]

# nominal data: randomly assign each doctor to a department
doc_dept_id = np.random.choice(department_id, no_of_drs)


# ordinal data: generating doctor grade in terms of senior, junior and consultant
grade = ['Senior Dr','Junior Dr','Consultant']
doctor_grade = np.random.choice(grade, no_of_drs)

# create DataFrame
doctor_data = pd.DataFrame({
    'Doctor_ID':doc_id,
    'Dept_Id':doc_dept_id,
    'Doctor_Name':doctors_names,
    'Doctor_Grade':doctor_grade
})

print(doctor_data.head(10))

```

```python
# number of Appointments

n = 1500

# nominal data: generate unique ID for each appointment
appointment_id = [f'APP{str(i).zfill(3)}' for i in range(1, 1501)]

# patient id is randomly assigned to each appointment - nominal data
app_patient_id = np.random.choice(Patient_id, n)

# doctor id is randomly assigned to each appointment - nominal data
app_doctor_id = np.random.choice(doc_id, n)

# interval data: generate random appointment date in years, month and days
app_year = np.random.randint(2005, 2026, n)
app_month = np.random.randint(1, 13, n)
app_day = np.random.randint(1, 31, n)
appointment_date = [f'{app_year[i]}-{str(app_month[i]).zfill(2)}-'
                     f'{str(app_day[i]).zfill(2)}' for i in range(n)]

# interval data: patirnt temp randomly generated in celsius, rounded to one d.p
patient_temp_celsius = np.round(np.random.uniform(32.0, 56.0, n), 1)

# created missing values for the aim of this assignment
# generate random waiting times between 5 and 120 minutes - ratio data
waiting_time_mins = np.random.randint(5, 121, n).astype(float)
# Randomly select 60 indices to set to NaN
n_points = 60
random_indices = np.random.choice(waiting_time_mins.size, n_points, replace=False)
waiting_time_mins[random_indices] = np.nan

# ordinal data: priority was assigned to appointment
priority = ['Routine', 'Soon','Urgent','Emergency']
priority_range = np.random.choice(priority, n, p=[0.4, 0.1, 0.2,0.3])

# nominal data: status was assigned to appointment to check if it has been
# completed, cancelled, rescheduled, or not available
status = ['Completed','Cancelled','Rescheduled','Not available']
appointment_status = np.random.choice(status, n, p=[0.2,0.4,0.15,0.25])

# consultation fee in pounds was generated between 20 - 200 pounds
# with missing values inclusive and rounded to two d.p - ratio data
consultation_fee_gbp = np.round(np.random.uniform(20, 200, n), 2)
# Randomly select 75 indices to set to NaN
n_points = 75
random_indices = np.random.choice(consultation_fee_gbp.size, n_points, replace=False).astype(int)
consultation_fee_gbp[random_indices] = np.nan

# create the DataFrame
Appointment_data = pd.DataFrame({
    'Appointment_id':appointment_id,
    'Patient_id':app_patient_id,
    'Doctor_id':app_doctor_id,
    'Appointment_date':appointment_date,
    'Patient_temp':patient_temp_celsius,
    'Waiting_time':waiting_time_mins,
    'Priority_Grade':priority_range,
    'Patient_status':appointment_status,
    'Consultation_fee_GBP':consultation_fee_gbp
})



print(Appointment_data.head(10))
```

```python
# number of servives
n = 30

# Nominal data
# generating  unique service id 
service_id = [f'SV{str(i).zfill(2)}' for i in range(1, n+1)]

# nominal data: service label
# defining service names and attaching them as a label 
svc_names = ['General Consultation','Blood Test','MRI Scan','Dental Care',
             'ECG','Ultrasound','Immunisation','X-ray']
service_label = np.random.choice(svc_names, n)

# nominal data: service_category 
service_cat = np.random.choice(['Diagnostic','Treatment','Preventive'
                                'Emergency'], n)

# Ordinal data: service rating ranging from 1 to 5
service_rating = np.random.randint(1, 6, n)

# nominal data: service description using faker to generate 
# the sentences describing each service
service_description = [fake.sentence(nb_words=9) for i in range(n)]

# ratio: service cost in pounds
service_cost_GBP = np.round(np.random.uniform(40, 1000, n), 2)

# create services DataFrame
service_data = pd.DataFrame({
    'Service_ID':service_id,
    'Service_names':service_label,
    'Service_category':service_cat,
    'Service_rating':service_rating,
    'Service_description':service_description,
    'Service_cost_GBP':service_cost_GBP
})

print(service_data.head(5))



```

```python
# Number of service appointment records
n = 600

# generate unique appointment service id
# nominal data: appointment service id
appointment_service_id = [f'APS{str(i).zfill(4)}' for i in range(1, n+1)]

# nominal data: this randomly selects appointment id from the appointment data table
app_service_appointment_id = np.random.choice(Appointment_data['Appointment_id'], n)

# nominal data: this randomly selects service ids from service data table
app_service_id = np.random.choice(service_data['Service_ID'], n)

# ratio data: service quantity in terms of how many services the patient opted for
service_qty = np.ones(n, dtype=int)

# ratio data: this randomly selects service cost from the service data table
app_service_cost = np.random.choice(service_data['Service_cost_GBP'],n)

# create the DataFrame for Appointment service
Appointment_service_data = pd.DataFrame({
    'Appointment_Service_id':appointment_service_id,
    'Appointment_ID':app_service_appointment_id,
    'Service_ID':app_service_id,
    'Service_Quantity':service_qty,
    'Service_Cost_GBP':app_service_cost
})

# created a few duplicate rows in the data to give a realistic frame of the database
# the duplicated rows were appended with the original dataframe
duplicate_row = Appointment_service_data.sample(n=5, random_state=36)
Appointment_service_data = pd.concat([Appointment_service_data, duplicate_row], ignore_index=True)

# regenerated appointment service id 
Appointment_service_data['Appointment_Service_id'] = [f'APS{str(i).zfill(4)}' for i in range(1, len(Appointment_service_data) +1)]


print(Appointment_service_data.head(5))


```

```python
# set the index of each data
Patient_data.set_index('Patient_ID', inplace=True)
department_data.set_index('Dept_ID', inplace=True)
doctor_data.set_index('Doctor_ID', inplace=True)
Appointment_data.set_index('Appointment_id', inplace=True)
service_data.set_index('Service_ID', inplace=True)
Appointment_service_data.set_index('Appointment_Service_id', inplace=True)
```

```python
# splitting multiple pandas DataFrame and saving the tables as csv files
Patient_data.to_csv('Patient_Data.csv')
department_data.to_csv('Department_data.csv')
doctor_data.to_csv('Doctor_Data.csv')
Appointment_data.to_csv('Appointment_Data.csv')
service_data.to_csv('Service_Data.csv')
# reset the index so appointment_service_id becomes a regular column and
# csv is saved without an additional generated index column
Appointment_service_data.reset_index().to_csv('Appointment_Service_Data.csv',index=False)
```
